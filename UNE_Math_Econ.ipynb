```mermaid
flowchart LR
    LA["$$\begin{matrix} R^n \Rightarrow R^m (A_{m \times n}) \end{matrix}$$"]
    
    LA --> up[m > n <br> dim rising<br> 單射可能]
    LA --> sq[m = n <br> **square matrix** <br> has det]
%% 2. 方陣的全局性質與 det (宏觀行為)
    sq -->|injection| inj[rank r=n <br> 零空間為零]
    sq -->|surjection| surj[r=m <br> 像空間為全空間]
    sq -->|bijection<br> **det** not 0| bij["r=m=n <br>A⁻¹ exists" ]
    sq -->|Singular/降秩<br>**det = 0** <br> at least 1 eigen=0| sing[ r \< n <br> kernel not 0 ]
    sq -->|Special matrices| spe[投影矩陣/冪等降維 <br> symm/orthog/diag/Hermitian]

    LA --> low[m \< n <br> dim reduction <br>projection/compression]


- 冪等矩陣（Idempotent Matrix）嚴格來說只能是方陣（Square Matrix）. 如果你是用非方陣的投影算子$A_{m \times n}$ 階，將低維度投影到高維度或反之, 來構造投影公式 $P = A(A^T A)^{-1} A^T$，最後算出來的結果 $P$ 仍然是一個 $m \times m$ 的方陣


```mermaid
flowchart LR
    LA["$$\begin{matrix} R^n \Rightarrow R^m (A_{m \times n}) \end{matrix}$$"]
    
%% 4. 通用內在結構 (SVD)
    LA -->|數值計算或理解正交化| qr[QR decomposition]
    LA -->|方陣且需要理解動態系統/長期行為| eig[eigen decomposition，尋找不變方向（特徵向量）和增長率/縮放比例（特徵值]
    LA -->|極值行為<br>最大拉伸、最小拉伸| svd["SVD for all m×n <br> 旋轉→拉伸→旋轉"]
    svd -.->|"對稱矩陣特化: <br> 奇異值U=|eigenV|"| sym[sym mat:主軸拉伸]
    svd -.->|正交矩陣特化: <br> 所有奇異值=1| orth[orth mat: 剛體運動]

### Econ

經濟學和商業相關課程屬於 **HSIE（人類社會及其環境）** 學習領域。

1. 商業選修課程：Commerce（7–10年級）通常不會將經濟和商業單獨拆成獨立學科，而是統一合併在 **Commerce（商學/商業）** 這門選修課中。
    * **主要修讀年級**：**9年級和10年級**（Stage 5）。雖然新版Syllabus在定義上涵蓋了 7–10 年級（Stage 4 & Stage 5），但NSW絕大多數中學都是將其作為 **9–10年級的兩年制選修課**（可選擇100小時或200小時課程）。
    * **核心內容**：涵蓋消費者與金融環境、個人理財、經濟環境（通膨、失業、市場運作）、商業與創業環境（如何營運公司、市場行銷）、法律與政治環境，以及工作環境。
    * **定位**：這門課是高中高年級經濟學和商業研究（Business Studies）的重要奠基課程。

2. 高中高級課程：HSC 階段（11–12年級）商科會正式分流為獨立的專業學科，學生可以根據興趣自由選修：
- **經濟學（Economics 11–12）**
    * **11年級（Preliminary 階段）**：微觀經濟（Microeconomics）焦點。主要學習經濟學基礎原理、市場運作（需求與供給）、消費者與企業行為、金融市場、政府在經濟中的角色，以及初步的國際經濟概念。
    * **12年級（HSC 階段）**：宏觀經濟（Macroeconomics）焦點。核心探討全球經濟（Global Economy）、澳洲經濟現狀與面臨的議題（如經濟增長、失業、通膨、財富分配），以及政府如何運用財政政策和貨幣政策進行經濟管理。
- **商業研究（Business Studies 11–12）**  | 獨立的高級 HSC 選修學科 |
    * **11年級**：探討商業的本質、中小企業（SMEs）的建立與營運，以及商業計劃。
    * **12年級**：深入學習大型企業的核心四大職能：營運（Operations）、行銷（Marketing）、財務（Finance）與人力資源（Human Resources）。

| 年級 (Years) | 對應階段 (Stages) | 核心學科名稱 | 課程性質 |
| --- | --- | --- | --- |
| **7–8 年級** | Stage 4 | 無獨立商科 | 僅在歷史/地理或學校特設的極短期選修中涉及基礎概念 |
| **9–10 年級** | Stage 5 | **Commerce** (商學) | 熱門選修課（內含經濟與商業基礎） |
| **11–12 年級** | Stage 6 (HSC) | **Economics** (經濟學)<br>

> *備註：新標準（2024版 Commerce 綱要與 2025版 Economics 綱要）將陸續於 **2027年** 全面強制實施，目前的教學與排課仍遵循 9-10 年級選修 Commerce、11-12 年級分流經濟與商業的成熟架構。*

### PMTH213 - Linear Algebra

- Assessment 20% - 5 times
- Final Exam 80%

linear transformations between vector spaces. This unit offers you an introduction to linear algebra using familiar examples, and explores earlier results, like the Fundamental Theorem of Calculus, in a new light. Building on your understanding of algebra, DEs and multivariable calculus, you will explore systems of linear equations, matrices and algebraic operations on them, and inner products. Learning outcomes:
1. unify algebra, DEs and multivariate calculus; 不要孤立地學習矩陣。特徵值立刻去複習二階常係數 ODE 的求解；雅可比矩陣去複習多元微積分中的鏈式法則和極值問題。**建立「萬物皆可線性化」的思維**。
2. theoretical understanding and computational methods; 考試通常分為兩部分。純計算, 如求 SVD、做 Gram-Schmidt; 另一部分是證明（如證明某個映射是線性變換、證明子空間的性質）。
3. prepare for later studies in algebra, statistics, functional analysis, complex variables, differential geometry, signal processing, harmonic analysis, physics;  理解「正交投影」在幾何和數據擬合中的雙重意義。
4. illustrate, demonstrate and teach modern mathematics. 「費曼技巧」。在複習時，嘗試用直觀的幾何圖形（如二維/三維空間中的拉伸、旋轉、投影）來解釋抽象的代數概念（如特徵值、核、像）。**特別關注「微積分算子」的線性代數解釋**，這是這門課拿高分和真正領悟數學之美的關鍵。

1. 矩陣代數與線性方程組（具體與計算基礎）*對應大綱：familiar examples, systems of linear equations, matrices and algebraic operations*

*   **線性方程組與矩陣運算：**
    *   高斯消元法（Gaussian elimination）與高斯-若爾當消元法。
    *   矩陣的代數運算（加法、乘法、轉置、逆矩陣）。
    *   行列式（Determinants）的性質與計算（克萊姆法則）。
*   **矩陣的分解與標準型（計算方法）：**
    *   LU 分解（簡介，為後續數值計算做準備）。
    *   矩陣的秩（Rank）、零度（Nullity）與秩-零化度定理（Rank-Nullity Theorem）。

2. 抽象向量空間與線性變換（核心理論）*對應大綱：linear transformations between vector spaces, introduction using familiar examples*

*   **向量空間的公理化定義：**
    *   從熟悉的 $\mathbb{R}^n$ 過渡到抽象向量空間。
    *   **非矩陣向量空間的例子：** 多項式空間 $P_n$、連續函數空間 $C[a,b]$、矩陣空間 $M_{m \times n}$。
*   **子空間與基：**
    *   子空間（Subspaces）、張成（Span）、線性相關與無關。
    *   基（Basis）與維度（Dimension）。
*   **線性變換（Linear Transformations）：**
    *   線性變換的定義、核（Kernel）與像（Image）。
    *   **線性變換的矩陣表示：** 如何在給定基底下將抽象線性變換轉化為矩陣。
    *   基變換（Change of basis）與過渡矩陣（Transition matrices）。
    *   同構（Isomorphism）：證明有限維向量空間與其坐標空間 $\mathbb{R}^n$ 同構。

3. 特徵值理論與微分方程的聯繫（統一性與應用）*對應大綱：unify material learnt in earlier units (DEs), computational methods*

*   **特徵值與特徵向量：**
    *   特徵多項式、特徵值、特徵空間。
    *   矩陣的對角化（Diagonalization）條件與方法。
    *   凱萊-哈密頓定理（Cayley-Hamilton Theorem）。
*   **線性代數在常微分方程（DEs）中的統一應用：**
    *   將高階線性 ODE 轉化為**一階線性 ODE 方程組（狀態空間形式）**。
    *   利用矩陣指數（Matrix exponential, $e^{At}$）和特徵值/特徵向量求解線性常係數 ODE 方程組。

4. 內積空間與正交性（通往高級課程的橋樑）*對應大綱：inner products, prepare for statistics, signal processing, harmonic analysis*

*   **內積空間基礎：**
    *   實/複內積的定義與性質。
    *   範數（Norm）、距離、柯西-施瓦茨不等式（Cauchy-Schwarz Inequality）。
*   **正交性與投影：**
    *   正交補（Orthogonal complements）。
    *   **格拉姆-施密特正交化過程（Gram-Schmidt process）**。
    *   正交投影（Orthogonal projections）與**最小二乘法（Least Squares approximations）**（統計學與數據擬合的核心）。
*   **正交矩陣與酉矩陣：**
    *   正交矩陣（Orthogonal matrices）的性質。
    *   實對稱矩陣的對角化（譜定理 Spectral Theorem 基礎）。

5. 微積分與線性代數的統一（核心亮點）*對應大綱：explores earlier results, like the Fundamental Theorem of Calculus, in a new light; unify multivariable calculus*

*   **微積分算子作為線性變換：**
    *   將微分算子 $D = \frac{d}{dx}$ 和積分算子 $I = \int$ 視為函數空間（如多項式空間 $P_n$ 或光滑函數空間）上的**線性變換**。
    *   求微分算子的矩陣表示（例如在基 $\{1, x, x^2\}$ 下）。
*   **用線性代數重新詮釋微積分基本定理（FTC）：**
    *   從線性代數角度看，FTC 本質上是關於**核（Kernel）與像（Image）** 的關係。
    *   微分算子 $D$ 的核是常數空間。積分算子 $I$ 是 $D$ 的右逆（$D \circ I = \text{Id}$），而 $I \circ D = \text{Id} - P_0$（其中 $P_0$ 是投影到常數空間的投影算子）。這完美解釋了為什麼不定積分要加常數 $C$。
*   **多變量微積分中的線性代數：**
    *   **雅可比矩陣（Jacobian matrix）** 作為多變量函數在某一點的「最佳線性近似（Best linear approximation）」。
    *   海森矩陣（Hessian matrix）與多元函數的極值判定（正定/負定矩陣）。

6. 高級主題與數學表達 *對應大綱：prepare for functional analysis, differential geometry, physics; illustrate and teach modern mathematics*

*   **對偶空間與張量初步（為微分幾何/物理準備）：**
    *   線性泛函（Linear functionals）與對偶空間（Dual space, $V^*$）。
    *   伴隨變換（Adjoint transformations）。
*   **奇異值分解（SVD）與主成分分析（PCA）初步（為統計/信號處理準備）：**
    *   SVD 的幾何意義與計算。
    *   協方差矩陣（Covariance matrices）與數據降維的線性代數基礎。
*   **數學溝通與證明（LO4 的落實）：**
    *   嚴格的數學證明寫作（直接證明、反證法、數學歸納法在線性代數中的應用）。
    *   可能包含小組展示（Presentation）或撰寫數學解釋報告，練習如何「illustrate and teach」數學概念。


### PMTH212 - Multivariable Calculus

basic concepts of single variable calculus, generalising these concepts to functions of two or more variables. Focusing on both theoretical and technical aspects, you will explore basic geometrical topics on curves and surfaces in relation to multivariable functions. You will also examine limits and continuity, differentiability and partial derivatives, inverse and implicit mapping theorems, multivariable Taylor formula, extreme values, double and triple integrals, line integrals and the integral theorems. Applying your knowledge to fluid dynamics and using the principles of multivariable calculus, the unit will help you to fine tune your ability to solve complex real-world mathematical problems. 核心难点在于**从一维的“线”思维跨越到三维的“体/面”思维**，以及**向量场三大定理的综合运用**。掌握坐标变换下的积分计算，并深刻理解散度/旋度在流体力学中的物理图像，是拿下这门课高分的关键。

Learning outcomes: - theoretical and technical knowledge
1. Vector functions, length, curvature:
    *   **策略：** 这一部分计算量较大且容易出错。重点掌握**弧长参数化**的技巧，因为一旦曲线用弧长参数化，曲率和 Frenet 标架的计算公式会大幅简化。务必熟记 $\mathbf{T}, \mathbf{N}, \mathbf{B}$ 的相互关系（如 $\mathbf{B} = \mathbf{T} \times \mathbf{N}$）。
2.  Differentiation, Inverse/Implicit mapping, Taylor, Extrema:
    *   **策略：** 区分“计算”与“理论”。计算部分（偏导、梯度、拉格朗日乘数法）必须做到熟练无误；**理论部分（逆映射/隐映射定理）是拉开分差的关键**。不要死记硬背定理，要理解其几何意义（例如：雅可比行列式不为零意味着局部映射是“非退化”的，即没有发生“折叠”）。理解多变量泰勒公式中 Hessian 矩阵的正定/负定如何决定极值性质。
3.  Integration, surface and volume integrals:
    *   **策略：** 积分部分的难点在于**画图与确定积分限**。在计算三重积分时，必须熟练掌握直角、圆柱、球坐标的切换（根据区域的对称性选择坐标系）。对于第二类曲面积分（通量），**法向量的方向（朝向）极其重要**，务必根据题目要求（如“向外”）正确确定法向量的正负号。
4.  Integral theorems & Fluid dynamics:
    *   **策略：** 这是多变量微积分的“大boss”。
        *   首先，必须能够**在三个积分定理之间灵活切换**。例如，计算一个复杂的曲面积分，如果曲面是封闭的，优先考虑散度定理转化为体积分；如果曲面有边界，考虑斯托克斯定理转化为线积分。
        *   其次，**死磕流体应用**。在考试中，应用题通常会给出一个速度场 $\mathbf{v}$，要求你解释 $\nabla \cdot \mathbf{v}$ 和 $\nabla \times \mathbf{v}$ 的物理意义，并计算特定区域的流量或环量。把“散度=膨胀率/源”，“旋度=旋转率/涡度”刻在脑子里。

强调的“理论与技术并重（theoretical and technical aspects）”、“逆映射与隐映射定理”以及“应用于流体动力学（fluid dynamics）”可以看出，这门课**不仅仅是大一单变量微积分的简单计算延伸**，而是引入了更严格的数学分析基础，并高度强调向量微积分在物理（特别是流体力学）中的实际应用。

1. 空间解析几何与向量值函数 vector functions and curves in space, length and curvature, basic geometrical topics on curves and surfaces

*   **三维空间与向量基础：**
    *   三维空间中的向量运算、直线与平面方程。
*   **向量值函数（Vector-valued functions）：**
    *   空间曲线的参数化（Parameterization）。
    *   向量函数的极限、连续性、导数与积分。
*   **曲线的几何性质（核心计算）：**
    *   **弧长（Arc length）** 的计算与弧长参数化（Arc-length parameterization）。
    *   **曲率（Curvature）** 与曲率半径的计算。
    *   **Frenet 标架（Frenet Frame）：** 单位切向量（$\mathbf{T}$）、主法向量（$\mathbf{N}$）、副法向量（$\mathbf{B}$）的计算及其几何意义。
    *   挠率（Torsion）简介（视具体教学深度而定）。

2. 多变量微分学与高级定理 limits and continuity, differentiability, partial derivatives, inverse and implicit mapping theorems, multivariable Taylor formula, extreme values

*   **多变量函数的基础微分：**
    *   多变量函数的极限与连续性（可能涉及严格的 $\epsilon-\delta$ 定义或路径极限）。
    *   偏导数（Partial derivatives）与**方向导数（Directional derivatives）**。
    *   梯度（Gradient）及其几何意义（等高线的法向量）。
    *   **可微性（Differentiability）** 的严格定义与全微分（Total differential）。
    *   多变量链式法则（Chain rule）与高阶偏导数（Clairaut's Theorem）。
*   **高级映射定理（理论核心）：**
    *   雅可比矩阵（Jacobian matrix）与雅可比行列式。
    *   **逆映射定理（Inverse Mapping Theorem）：** 局部可逆的条件与雅可比行列式的关系。
    *   **隐映射定理（Implicit Mapping Theorem）：** 隐函数存在定理及其在求导和几何（如曲面切平面）中的应用。
*   **近似与极值（技术与应用）：**
    *   **多变量泰勒公式（Multivariable Taylor formula）：** 二次泰勒多项式与海森矩阵（Hessian matrix）。
    *   无约束极值与二阶导数判别法。
    *   **有约束极值与拉格朗日乘数法（Lagrange multipliers）**（重点难点）。

3. 多变量积分学 double and triple integrals, compute surface and volume integrals

*   **重积分基础：**
    *   二重积分（Double integrals）的定义、性质与富比尼定理（Fubini's Theorem）。
    *   三重积分（Triple integrals）的计算。
    *   **坐标变换：** 极坐标（Polar）、圆柱坐标（Cylindrical）与球坐标（Spherical coordinates）下的积分计算。
*   **曲线积分（Line Integrals）：**
    *   第一类曲线积分（关于弧长的积分，标量场）。
    *   第二类曲线积分（关于坐标的积分，向量场做功）。
*   **曲面积分（Surface Integrals）：**
    *   曲面的参数化与法向量的计算。
    *   第一类曲面积分（标量场在曲面上的积分，如质量）。
    *   **第二类曲面积分（向量场在曲面上的积分 / 通量 Flux integrals）**（核心难点）。

4. 向量微积分核心定理与流体应用 integral theorems, apply this knowledge to fluid dynamics

*   **向量场分析：**
    *   散度（Divergence, $\nabla \cdot \mathbf{F}$）与旋度（Curl, $\nabla \times \mathbf{F}$）的计算与物理意义。
    *   保守场（Conservative fields）、势函数（Potential functions）与路径无关性。
*   **三大核心积分定理（理论巅峰）：**
    *   **格林定理（Green's Theorem）：** 平面区域上的线积分与二重积分的关系。
    *   **斯托克斯定理（Stokes' Theorem）：** 空间曲面上的线积分与曲面积分的关系。
    *   **散度定理 / 高斯定理（Divergence Theorem）：** 封闭曲面上的通量与体积分的关系。
*   **流体动力学应用（UNE 课程特色）：**
    *   将流体速度场 $\mathbf{v}(x,y,z)$ 建模为向量场。
    *   **流量（Flux）与散度：** 计算流体穿过曲面的流量，利用散度定理分析流体的“源（Sources）”与“汇（Sinks）”。
    *   **环量（Circulation）与旋度：** 计算流体沿闭合曲线的环量，利用斯托克斯定理分析流体的“涡度（Vorticity）”与旋转特性。
    *   连续性方程（Continuity equation / 质量守恒）的积分形式推导。


### PMTH339 - Differential Equations

- Assessment 20% - 9 times
- Final Exam 80%

Offers you qualitative and quantitative methods for ODE and PDE. You will explore first order equations, second order linear equations, series solutions, boundary value problems and phase plane analysis. Learning outcomes:

1. use a broad and coherent theoretical knowledge to demonstrate an in-depth understanding of the theory of elementary, ODE and PDE and some of its applications;
    *   **重點：** 不要只背公式。要理解「為什麼」解是這個形式。例如，理解 Wronskian 為什麼能判斷線性無關，理解特徵值問題為什麼在 ODE 和 PDE 中反覆出現。
2. develop a deep level of theoretical knowledge about PDE;
    *   **重點：** 這是 3 年級課程的難點。必須熟練掌握 **Sturm-Liouville 理論** 和 **PDE 的分類與特徵線**。理解邊界條件（Boundary Conditions）如何決定解的性質（例如傅里葉係數的計算）。
3. apply well-developed knowledge and skills relating to applications of DEs; and
    *   **重點：** 考試中通常會有 Application 大題。你需要熟練掌握「文字題 -> 數學方程 -> 求解 -> 解釋物理意義」的完整流程。特別注意相平面分析在生物模型中的應用。
4. analyse and generate solutions to sometimes complex problems, with logical and coherent methodolgy in solving DEs.
    *   **重點：** 培養「解題框架」思維。遇到複雜方程時，第一步判斷類型（線性/非線性、常/偏、階數），第二步選擇合適的方法（分離變量、級數、積分變換、定性分析），第三步執行計算，第四步驗證與解釋。

1. 一階常微分方程與定性分析 (First-Order ODEs & Qualitative Analysis) *對應大綱：first order equations, qualitative methods, phase plane analysis*

*   **定量求解方法（Quantitative Methods）：**
    *   可分離變量方程、齊次方程。
    *   一階線性方程與積分因子（Integrating factors）。
    *   全微分方程（Exact equations）與伯努利方程（Bernoulli equations）。
*   **存在性與唯一性理論（Theoretical Knowledge）：**
    *   皮卡-林德洛夫定理（Picard-Lindelöf Theorem）與皮卡迭代法。
    *   解對初值和參數的連續依賴性。
*   **定性分析與相平面分析（Qualitative & Phase Plane Analysis）：**
    *   方向場（Direction fields）與等斜線（Isoclines）。
    *   自治系統（Autonomous systems）與一維相線分析。
    *   **二維相平面分析：** 臨界點/平衡點（Equilibrium points）的分類（節點、鞍點、螺旋點、中心點）。
    *   非線性系統的線性化（Linearization）與李雅普諾夫穩定性（Lyapunov stability）基礎。
    *   極限環（Limit cycles）與龐加萊-本迪克松定理（Poincaré-Bendixson Theorem）簡介。

2. 高階線性常微分方程與級數解 (Higher-Order Linear ODEs & Series Solutions) *對應大綱：second order linear equations, series solutions*

*   **二階及高階線性 ODE 理論：**
    *   齊次與非齊次方程的解的結構。
    *   朗斯基行列式（Wronskian）與阿貝爾定理（Abel's Theorem）。
    *   常係數線性方程的求解（特徵方程法）。
    *   待定係數法（Method of Undetermined Coefficients）與參數變易法（Variation of Parameters）。
*   **冪級數解法（Series Solutions）：**
    *   常點（Ordinary points）處的冪級數解。
    *   正則奇點（Regular singular points）處的 **Frobenius 方法**。
    *   指標方程（Indicial equation）與根的分類（實根、重根、複根）對應的解的形式。
*   **特殊函數（Special Functions - 3年級核心考點）：**
    *   貝塞爾方程（Bessel's equation）與貝塞爾函數。
    *   勒讓德方程（Legendre's equation）與勒讓德多項式。

3. 邊值問題與施圖姆-劉維爾理論 (Boundary Value Problems & Sturm-Liouville Theory) *對應大綱：boundary value problems, 連接 PDE 的橋樑*

*   **邊值問題（BVPs）基礎：**
    *   初值問題（IVP）與邊值問題（BVP）的區別（解的存在性與唯一性差異）。
*   **施圖姆-劉維爾理論（Sturm-Liouville Theory）：**
    *   正則施圖姆-劉維爾問題（Regular S-L problems）的標準形式。
    *   特徵值（Eigenvalues）與特徵函數（Eigenfunctions）的性質。
    *   特徵函數的正交性（Orthogonality）與權函數（Weight functions）。
*   **傅里葉級數（Fourier Series）：**
    *   傅里葉正弦級數與餘弦級數。
    *   週期函數的傅里葉展開與狄利克雷條件（Dirichlet conditions）。

4. Deep Theory of PDEs *對應大綱：PDE, deep level of theoretical knowledge about PDE*

*   **PDE 基礎與分類：**
    *   二階線性 PDE 的分類：橢圓型（Elliptic）、拋物型（Parabolic）、雙曲型（Hyperbolic）。
    *   特徵線方法（Method of Characteristics）與標準型化簡。
*   **分離變量法（Separation of Variables）：**
    *   將 PDE 轉化為 ODE 邊值問題的系統方法。
*   **三大經典 PDE 的解析解：**
    *   **熱傳導方程（Heat/Diffusion Equation）：** 一維/二維熱傳導，穩態解。
    *   **波動方程（Wave Equation）：** 達朗貝爾解（D'Alembert's solution），行波與駐波，弦振動問題。
    *   **拉普拉斯/泊松方程（Laplace's/Poisson's Equation）：** 穩態溫度分佈，極值原理（Maximum principle）。
*   **積分變換法（Integral Transforms）：**
    *   拉普拉斯變換（Laplace Transform）在求解 PDE 和 ODE 中的應用。
    *   傅里葉變換（Fourier Transform）在無限域 PDE 問題中的應用。

5. Applications & Complex Problem Solving *對應大綱：applications of DEs, analyse and generate solutions to complex problems*

*   **物理與工程建模：**
    *   機械振動（彈簧-質量-阻尼系統，共振現象）。
    *   電路分析（RLC 電路）。
    *   連續介質力學基礎（弦的橫振動、桿的縱振動推導）。
*   **生物與生態建模（結合定性分析）：**
    *   種群動力學（Population dynamics）。
    *   Lotka-Volterra 捕食者-獵物模型（利用相平面分析其週期性與穩定性）。
*   **複雜問題的方法論（Logical and coherent methodology）：**
    *   如何將複雜的實際物理/工程問題轉化為數學模型（建立 DE）。
    *   當精確解析解無法求得時，如何結合定性分析（如穩定性）和近似/漸近方法來理解系統行為。

### Garage

PMTH332 - Abstract Algebra

This unit will introduce you to the theory of groups and rings. You will explore basic properties, subgroups and subrings, quotient structures, products of groups and rings, and isomorphism theorems. You will meet key results such as the Fundamental Theorem of Finitely Generated Abelian Groups and the Sylow Theorems as well as cyclic groups, polynomial rings and fields of quotients. To boost your ability to construct proofs and your communication of complex concepts to prepare for further studies in advanced mathematics. Learning outcomes:

1. demonstrate an understanding of basic topics in Abstract Algebra;
2. demonstrate a preliminary understanding of central elementary concepts of modern algebra; 抽象代數的定義極多且極其嚴格。**必須做到對定義的「肌肉記憶」**。例如，證明一個子集是正規子群，必須嚴格寫出 $gNg^{-1} \subseteq N$ 的過程，不能跳步。
3. continue the unification of earlier material commenced in PMTH213; 在學習「商群/商環」和「同態」時，不斷回想線性代數中的「商空間（Quotient space）」和「線性映射的核與像（Rank-Nullity Theorem）」。你會發現抽象代數的定理是線性代數定理在更一般結構下的完美推廣。
4. develop both theoretical understanding and computational methods;    *計算題：* 熟練計算對稱群的置換、求商群的階、利用西羅定理計算特定階數群的西羅子群數量、將有限阿貝爾群分解為標準型。 *證明題：* 掌握「構造同構映射」的標準套路（定義映射 -> 證明良定義 -> 證明同態 -> 證明單射 -> 證明滿射）。
5. prepare themselves for further studies in number theory, commutative algebra, Lie algebras, algebraic geometry, representation theory, algebraic topology, homological algebra, category theory;  在學到「多項式環」和「極大理想」時，主動去了解一下它們在代數幾何（Algebraic Geometry）中代表什麼（即「點」）；在學到「群作用」時，思考它在拓撲學（基本群）中的意義。這會極大拓寬你的數學視野。
6. illustrate, demonstrate and teach modern mathematics. 用**費曼技巧**：能不能用一句話或幾何/物理直觀例子（如魔方、撲克牌洗牌、晶體對稱性）解釋什麼是「群」或「商群」？

Contents:

**1. Group Theory & Cyclic Groups** *對應大綱：basic properties, subgroups, cyclic groups*
*   群的基本概念：
    *   群、阿貝爾群（交換群）的定義與基本性質。eg. 對稱群（Symmetric groups $S_n$）、二面體群（Dihedral groups $D_n$）、一般線性群/特殊線性群（矩陣群，這裡與 PMTH213 線性代數產生強烈連結）。
*   子群與循環群：
    *   子群（Subgroups）的判定定理。
    *   循環群（Cyclic groups）的生成元、階（Order）與分類（同構於 $\mathbb{Z}$ 或 $\mathbb{Z}_n$）。
*   陪集與拉格朗日定理：
    *   左/右陪集（Cosets）。
    *   拉格朗日定理（Lagrange's Theorem）及其推論（如費馬小定理、歐拉定理的群論證明）。

**2. 群的同態與高級結構** (Group Homomorphisms & Advanced Structures)
*對應大綱：quotient structures, products of groups, isomorphism theorems, Sylow Theorems, Fundamental Theorem of Finitely Generated Abelian Groups*
*   同態與商群：
    *   群同態（Homomorphisms）、核（Kernel）與像（Image）。
    *   正規子群（Normal subgroups）的定義與判定。
    *   商群（Quotient groups） 的構造與性質。
*   同構定理與直積：
    *   群的第一、第二、第三同構定理（核心考點）。
    *   群的直積（Direct products）與內部/外部直積。
*   有限群的高級分類定理（三年級核心難點）：群論-西羅定理和阿貝爾群分類定理，+環論-理想與商環構造，你將真正具備現代純數學的思維方式
    *   西羅定理（Sylow Theorems）： 西羅 $p$-子群的存在性、共軛性與數量計算（用於分類特定階數的群）。
    *   有限生成阿貝爾群基本定理（Fundamental Theorem of Finitely Generated Abelian Groups）： 將有限阿貝爾群分解為循環群直積的標準型（不變因子型與初等因子型）。

**3. Ring Theory** *對應大綱：rings, subrings, quotient structures, products of rings, isomorphism theorems*
*   環的基本概念：
    *   環、交換環、含幺環的定義。
    *   整環（Integral domains）與域（Fields）的定義與關係。
    *   子環（Subrings）與理想（Ideals：主理想、素理想、極大理想）。
*   環的同態與商環：
    *   環同態、核與像。
    *   商環（Quotient rings）的構造。
    *   環的同構定理（與群的同構定理對比學習）。
*   環的直積：
    *   環的直積結構與中國剩餘定理（Chinese Remainder Theorem）在環論中的推廣。

**4. 多項式環與商域** (Polynomial Rings & Fields of Quotients) *對應大綱：polynomial rings, fields of quotients*
*   多項式環理論：
    *   多項式環 $R[x]$ 的性質。
    *   帶餘除法、最大公因式。
    *   不可約多項式（Irreducible polynomials）與愛森斯坦判別法（Eisenstein's criterion）。
*   特殊整環的分類：
    *   歐幾里得整環（ED） $\subset$ 主理想整環（PID） $\subset$ 唯一分解整環（UFD） $\subset$ 整環（Integral Domain）。
    *   高斯引理（Gauss's Lemma）與多項式環的 UFD 性質。
*   商域（Fields of Quotients）：
    *   從整環構造其商域（例如從 $\mathbb{Z}$ 構造 $\mathbb{Q}$，從多項式環構造有理函數域）。
    *   商域的泛性質（Universal property）。

**5. 統一性與前瞻** (Unification & Forward Look) *對應大綱：continue the unification of earlier material commenced in PMTH213; prepare for further studies (number theory, algebraic geometry, Lie algebras, etc.)*
*   與 PMTH213（線性代數）的統一：
    *   模（Modules）初步： 將向量空間視為「域上的模」，將阿貝爾群視為「整數環 $\mathbb{Z}$ 上的模」。這將線性代數中的基、維度、線性變換統一到了抽象代數的框架下。
    *   矩陣環 $M_n(F)$ 的結構與線性變換的環同態。
*   為高級課程鋪路（LO5 的落實）：
    *   **數論：** 環論中的理想與數論中的整除性。
    *   **代數幾何：** 多項式環的極大理想與仿射空間中的點（希爾伯特零點定理預習）。
    *   **表示論/李代數：** 群作用（Group actions）與對稱性的代數化。
    *   **範疇論/同調代數：** 同態、核、像、同構定理本質上都是範疇論中「態射（morphisms）」的具體體現。

**6. 證明建構與數學溝通** (Proof Construction & Mathematical Communication) *對應大綱：boost your ability to construct proofs, communication of complex concepts, illustrate, demonstrate and teach*
*   抽象代數的證明技巧：
    *   構造性證明（如構造同構映射）。
    *   利用泛性質（Universal properties）進行證明。
    *   反證法與數學歸納法在群/環階數與整除性中的應用。
*   數學直觀與教學溝通（LO6 的核心）：
    *   幾何直觀化： 例如用正方形的對稱性（翻轉、旋轉）來「illustrate」二面體群 $D_4$；用時鐘算術來「demonstrate」商群 $\mathbb{Z}/n\mathbb{Z}$。
    *   數學寫作與演講： 課程可能包含書寫嚴格的證明作業（Proof assignments），甚至要求進行小組展示（Presentation），向同學「teach」某個抽象定理（如西羅定理的證明思路），訓練將極度抽象的概念轉化為他人能聽懂的語言。
