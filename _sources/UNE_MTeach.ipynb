- **建設性對齊 (Constructive Alignment)**
- 澳洲的教師認證非常強調 **"Data-driven decision making (數據驅動的決策)"**，
- 從 Unit Plan 到 GTPA ，**Evidence-based practice** 是貫穿所有產出層級的最高指導原則。
    - **Unit Plan** 解決了 **"教什麼 (What)"** 和 **"怎麼教 (How)"** 的問題；
    - **GTPA** 則是在 Unit Plan 的基礎上，加入了 **"對誰教 (Who - Context)"**、**"教得如何 (Impact - Assessing)"** 以及 **"為什麼這樣教 (Why - Reflecting)"** 的完整閉環。

第一層：課堂執行層 (Micro-level Implementation)
這是實習初期和日常教學中最基礎的產出，聚焦於單節課的落地。

*   **Lesson Plan (單節課教案)**：具體的 40-60 分鐘單節課教學流程。核心內容：
    *   **Learning Intentions & Success Criteria (學習意圖與成功標準)**：明確告知學生「這節課要學什麼」以及「如何算學會了」（常使用 WALT - We Are Learning To / WILF - What I'm Looking For）; 具體的提問策略 (Questioning strategies)；時間分配。
    *   **Curriculum Links (課綱連結)**：精準對接澳洲國家課綱 (ACARA) 或各州課綱 (如 VIC 的 VELS/VCAL, NSW 的 NESA) 的 Content Descriptors。
    *   **Lesson Sequence (教學流程)**：
        *   *Hook/Warm-up* (引起動機)
        *   *Explicit Teaching* (直接教學Direct Instruction/概念講解)
        *   *Guided/Independent Practice* (引導 / 獨立練習)
        *   *Plenary/Reflection* (總結與反思)
    *   **Formative Assessment (形成性評估工具)**：用於課堂即時反饋的工具，如 Exit tickets (出門條)、Mini-whiteboards (小白板) 記錄、觀察核對表 (Observation checklists)、Thumbs up/down 等
    *   **Teaching Resources & Materials (教學資源)**: 課件 (Slides)、工作紙 (Worksheets)、多媒體素材、實體教具等
  
第二層：單元設計層 (Meso - Unit Plan & Rubrics Design) 是連接單節課與宏觀課程的橋樑，也是 GTPA 中「Planning（規劃）」階段的核心產出。

*   **Unit Plan / Learning Sequence (單元教案/學習序列)**
    *   **定位**：通常包含 **3-5 節連貫的課**，圍繞一個特定的主題或技能展開。這是 GTPA 要求提交的核心文件。
    *   **完整內容包含**：
        1.  **Syllabus/Curriculum Links (課程大綱連結)**：明確對應澳洲國家課程 (Australian Curriculum) 或各州大綱的具體 Content Descriptions (內容描述) 和 Achievement Standards (成就標準)。
            *   **Overarching Inquiry Question (核心探究問題)**：引導整個單元的驅動問題。
            *   **Scope and Sequence (範圍與序列)**：將單元目標拆解到每週/每節課的進度表。
            *   **Cross-Curriculum Priorities (跨課程優先事項)**：必須融入澳洲課綱的三大核心（例如：原住民歷史與文化、澳洲與亞洲的連結、永續發展）。
        2.  **Contextual Considerations (背景考量)**：該班級的整體學情、 prior knowledge (先備知識)。
        3.  **Differentiation & Inclusion (差異化與融合教育)**：針對不同需求學生（如 EAL/D 英語非母語學生、GATS 原住民、Gifted & Talented 資優生、SEN 特殊教育需求學生）的具體調整策略 (Adjustments)。
        4.  **Teaching and Learning Sequence (教學序列)**：3-5 節課的宏觀流程概述，每節課的重點、教學法 (Pedagogy) 和使用的資源。
        5.  **Assessment Plan (評估計劃)**：明確指出哪一節課進行形成性評估，哪一節課進行總結性評估 (Summative Assessment)。
*   **Summative Assessment Task (總結性評估任務)**
    *   **內容**：單元結束時的正式評估（如測驗、 essay、項目報告、口頭演講）。
    *   **配套文件**：必須包含 **Marking Rubric / Criteria sheet (評分量規)**，明確區分 A/B/C/D 或 Excellent/Satisfactory/Needs Improvement 的具體表現標準。

第三層：背景與學情分析層 (Macro-level Context)
在進行任何教學設計前，必須對環境和對象進行深度剖析，這是 GTPA 的起點。

*   **School Context Profile (學校背景檔案)**
    *   **內容**：學校的社會經濟地位 (ICSEA value)、地理位置、辦學理念、特殊項目、社區資源等。
*   **Class Profile (班級檔案)**
    *   **內容**：班級人數、性別比例、整體學術水平、行為管理策略、班級動態。
*   **Focus Student Profiles (焦點學生檔案)**
    *   **內容**：GTPA 要求選取 **3 名焦點學生**（通常代表高、中、低不同學術水平，或包含特殊需求學生）。需詳細記錄他們的學習優勢、挑戰、興趣、Iep (個別化教育計劃，若有) 等。
*   **Differentiation & Inclusion Plan (差異化與融合教育設計)**：學生需求層 Equity. 核心內容：
    *   **EAL/D (英語為第二語言或方言) 學生**：提供鷹架 (Scaffolding)、視覺輔助、詞彙表。
    *   **NCCD (全國一致殘疾學生定義)**：針對自閉症、ADHD、學習障礙等學生的合理調整 (Reasonable Adjustments)，如延長考試時間、提供安靜空間。
    *   **Gifted and Talented (資優生)**：提供延伸任務 (Extension tasks) 或更深度的探究。
    *   **UDL (通用學習設計)**：確保教學素材具備多種呈現、表達和參與的方式。

第四層：證據收集與數據分析層 (Evidence & Data Analysis)
澳洲教師教育極度看重「用數據說話」，這一層級是證明教學有效性的關鍵。

*   **Student Work Samples (學生作品樣本)**
    *   **內容**：收集焦點學生在總結性評估任務中的實際作品（需隱去真實姓名，用 Student A/B/C 代替）。通常要求提供 Pre-assessment (前測) 和 Post-assessment (後測) 的對比。
*   **Assessment Data Records (評估數據記錄)**
    *   **內容**：全班成績分佈圖表、焦點學生的具體得分及在量規 (Rubric) 上的具體表現位置。
*   **Feedback Records (反饋記錄)**
    *   **內容**：教師提供給學生的書面或口頭反饋記錄，以及**學生對反饋的回應/改進證據**（這在 AITSL 標準中非常重要，證明反饋被使用了）。
*   **Reflective ePortfolio (反思電子作品集)**：理論與專業成長層. 蒐集大學期間的課程作業產出，並用學術理論進行反思。核心內容：
    *   **對標 APST (澳洲教師專業標準)**：必須針對 7 大標準、37 個細項（分為 Graduate, Proficient, Highly Accomplished, Lead 四個層級，畢業生需達到 **Graduate 層級**）提供證據（如教案、學生作業照片、教學影片）。
    *   **Critical Reflection (批判性反思)**：常用 Gibbs 或 Schön 的反思模型，探討「我為什麼這樣教？效果如何？如果重來我會怎麼做？這與Vygotsky或Piaget的理論有何關聯？」
  
第五層：終極綜合產出 —— GTPA (Graduate Teacher Performance Assessment)
**最後一次教育實習**期間必須完成一份 **綜合檔案袋 (Portfolio/Report)「證明你具備獨立執教能力」的完整數據與反思報告，GTPA 不及格，将无法完成学位。** NESA要求申请者必须达到 APST，才能申请教师注册（Teacher Registration）

考核形式与内容: 独立完成一个完整的“教学、学习与评估循环”（Teaching, Learning and Assessment Cycle）。 Portfolio 通常包含5个核心部分：

1. Context (情境脈絡分析)
*   **內容**：整合第三層的內容。分析實習學校的社區背景、學校數據（如 NAPLAN 成績、出勤率）、班級概況，以及 3 名焦點學生的詳細學情分析。
*   **班級分析**：詳細描述你所教班級的學生組成（EAL/D 比例、特殊需求學生、原住民學生、班級學習風氣）。
*   **目的**：證明你的後續教學設計是「基於學生數據與真實情境」而量身定制的，了解教學環境，並能識別學生的多樣性需求 (對應 AITSL 標準 1: Know students and how they learn)。

2. Planning (教學規劃)设计一个连贯的教学单元（通常包含 3 到 5 节课的教案），明确学习目标，并说明如何根据学生的不同背景和能力进行差异化教学。
*   **內容**：提交上述的 **Unit Plan (單元教案3-5節課)**。
*   **要求**：必須明確指出你如何根據第 1 點的「Context」來調整這個單元計畫。你如何設計評量？如何融入跨課程優先事項？
*   **目的**：證明實習生能根據課程大綱和學情，設計連貫、有邏輯且包含差異化教學的單元計劃 (對應 AITSL 標準 2: Know the content and how to teach it; 標準 3: Plan for and implement effective teaching and learning)。

3. Teaching (教學執行與證據)
*   **內容**：證明你真的在課堂上執行了這個計畫。
*   **產出**：
    *   教學影片片段（需取得家長/學校同意）。
    *   指導教師 (Mentor) 的觀課紀錄表。
    *   課堂上的師生互動照片、使用的教材、白板記錄。

4. Assessing (評量與數據分析) 收集至少 2-3 名不同学习水平（如高分、中等、需要帮助的学生）的**真实作业样本**，并给出你的批改和反馈。
*   **內容**：整合第四層的內容。展示學生的學習成果。這是 GTPA 最看重的部分之一：**你如何證明學生「真的學到了」？** —— **GTPA 的重中之重**
*   **產出**：
    *   **Pre-assessment (前測) 與 Post-assessment (後測) 數據對比**：用圖表展示全班學生的進步幅度。
    *   **Student Work Samples (學生作業樣本)**：挑選高、中、低 (High, Medium, Low) 三份學生作品，指出他們的理解誤區或優勢。
    *   **反饋與下一步**：記錄你給這 3 名學生的具體反饋**Feedback**（需符合 Hattie 的可見學習理論或 Growth Mindset 概念），並提出針對性的「下一步教學建議 (Next steps for learning)」。
    *   **Data Analysis**：分析哪些教學策略有效？哪些學生沒有達標？為什麼？
    *   **目的**：證明實習生能設計並使用有效的評估工具, 能解讀學生數據，並利用數據來改進教學 (對應 AITSL 標準 5: Assess, provide feedback and report on student learning)。

5. Reflecting (專業反思) 基于收集到的学生作业数据，分析自己的教学效果，反思哪些教学策略有效，哪些无效，并提出后续教学的调整方案。
*   **內容**：綜合上述整個單元的教學過程，写深度的學術反思报告。必須明確引用 AITSL Graduate Standards，舉例說明自己的教學如何影響了學生的學習，並提出自己未來作為新教師的專業發展方向 (對應 AITSL 標準 6: Engage in professional learning)。
*   **核心問題**：
    *   你的教學對學生的學習產生了什麼具體影響 (Impact on student learning)？
    *   作為一名新手教師，你發現了自己的哪些優勢與盲點？
    *   你未來的專業發展計畫 (Professional Learning Plan) 是什麼？你需要尋求哪些資深教師或專家的幫助？

给就读学生的建议
1.  **争取连续授课：** 确保你有机会连续教授一个完整的单元（3-5节课），而不是零散地代课，这样才能体现教学的连贯性和对学生学习的持续影响。
2.  **注意收集素材：** 在实习期间务必保留好学生的作业样本、评估数据和课堂照片，这些都是撰写 GTPA 报告不可或缺的核心证据。
    1.  **Professional Experience Report (實習評估報告)**：3 到 4 次實習（從觀察、助教到全職接課）。
    1.  **Lesson Observation Records / University Supervisor Report (聽課與授課觀察記錄 / 大學督導報告)**：由大學導師 (University Supervisor) 或學校指導老師 (Mentor Teacher) 填寫的觀察反饋表。
    1.  **Mentor Teacher Feedback (指導教師回饋)**：實習學校的帶班老師給予的日常回饋。
    1.  **Professional Experience Reports (實習成績單)**：指導老師根據 AITSL 標準對實習生進行的期末評分與評語。
    1.  **Professional Engagement (專業參與)**：參與教師會議、家長會、課外活動 (Extracurricular activities) 的紀錄。
    1.  **eAPe (electronic Portfolio of Evidence / 電子檔案袋)**：許多大學將教案、反思、學生作品、大學作業等上傳至此系統，直接映射到 AITSL 的 7 大領域標準，畢業時作為教師註冊的佐證材料。

wk4. Teaching Strategies: CL as a teaching strategy, Differentiation Graphic Organisers
- Cooperative Learning (CL)
    - CL is an instructional strategy in which learners work together in small groups. Collectively students contribute to the achievement of a common goal.
    - It is a specific classroom management and organising students for learning. Children have particular roles and therefore it is more groupwork. There is no room for free-riding.
- Johnson & Johnson 5 principles:
    - Positive Interdependence
    - Individual and Group Accountability
    - Promotive (Face-to-Face) Interaction
    - Interpersonal and Small-Group Skills
    - Group Processing

<img src="https://raw.githubusercontent.com/tomctang/NB_img/main/CL.png" alt="Causality-confounder" width="1000">

- CL Step-by-Step
    - Form Diverse Groups: Create groups of 3 to 4 students based on mixed abilities, backgrounds, and skill levels to maximize peer tutoring.
    - Assign Clear Roles: Give each student a specific job (e.g., Recorder, Timekeeper, Spokesperson, or Checker) to ensure everyone participates and stays accountable.
    - Structure the Task: Design a task that necessitates collaboration rather than just division of labor. Popular cooperative models include:
        - Think-Pair-Share: Students think independently on a prompt, discuss it in pairs, and share with the class.
        - Jigsaw Method: Each student in a group masters a specific sub-topic (an "expert"), then returns to their base group to teach what they learned to the others.
    - Teach Collaborative Skills: Before starting, explicitly model appropriate behaviors like active listening, taking turns, and resolving conflicts.

#### Module 2: Intellectual Quality


- wk4: the QTM dimension of Intellectual Quality, and the role in classroom management.
- wk5: continue to investigate different observational techniques - time sampling, and some different scales and checklists

wk4: **Intellectual Quality**
- is pedagogy focused on deep understanding of important, substantive concepts, skills and ideas. Such pedagogy treats knowledge as requiring active construction and engages students in higher-order thinking and communicating about what they are learning. (NSW DET, 2023).
- 6 elements: metalanguage, higher-order thinking, substantive communication, deep knowledge, deep understanding, and problematic knowledge


wk5: Methods

Two approaches
- Traditional approach: focus on the individual student and collect data on them
  - Time sampling: to measure behaviours using qualitative methods to record data. (just look for the target behaviour itself, eg. EVENT FREQUENCY RECORDING methods - Interval Sampling or Time Sampling)
- Classroom ecology approach: **self assessment scales** and an **ecology checklist** are informed by Classroom Ecology.

Observation techniques (eg. time sampling) and ecological assessment tools are complementary (Core Relationship: Data Triangulation)
- Establishing the Baseline (Checklists): Classroom Ecology Checklist (CEC) 
- Tracking the Impact: Time Sampling (TS)
- Action-Oriented Assessment: 
  - Self-assessment scales and CEC indicate what systems should be in place (e.g., "Are positive expectations clear?"), 
  - while TS reveals whether those systems are working (e.g., "How often is the student displaying positive engagement during the hour?")

- wk6: final dimension of the QTF, Significance

[Project-Based Learning at High-Tech High](https://www.youtube.com/watch?v=6rv_rmJYorE) breaks traditional Highschool molds by integrating arts, design, and PBL. The approach fosters a respectful, adult-like atmosphere that prepares students through real-world work.

wk6: **Significance**
- Intellectual Quality required the teacher to know their content in a deep and broad way. Significance dimension builds on from there and requires the teacher to know how the content fits in with what the students have done before, how it aligns with other learning areas the students might be engaged in, how it ties back with the students own lives. Doing this allows students to see the connections between what they are doing and the world around them. This makes learning relevant and significant to them.
- Significance 6 elements:
  - Background knowledge
  - Cultural knowledge
  - Knowledge integration
  - Inclusivity
  - Connectedness
  - Narrative

wk7: Duration and Event recording
- Duration recording is used to collect data on the amount of time a student spends engaging in a particular behaviour. 

- Event recording is used to collect data about the number of times a particular behaviour occurs. a specific time - for example for 30 minutes during a mathematics lesson
