## PMTH339 - Differential Equations

Offers you qualitative and quantitative methods for ODE and PDE. You will explore first order equations, second order linear equations, series solutions, boundary value problems and phase plane analysis. Learning outcomes:

1. use a broad and coherent theoretical knowledge to demonstrate an in-depth understanding of the theory of elementary, ODE and PDE and some of its applications;
    *   **重點：** 要理解「為什麼」解是這個形式。例如，理解 Wronskian 為什麼能判斷線性無關，理解特徵值問題為什麼在 ODE 和 PDE 中反覆出現。
2. develop a deep level of theoretical knowledge about PDE;
    *   **重點：** 難點。必須熟練掌握 **Sturm-Liouville 理論** 和 **PDE 的分類與特徵線**。理解邊界條件（Boundary Conditions）如何決定解的性質（例如傅里葉係數的計算）。
3. apply well-developed knowledge and skills relating to applications of DEs; and
    *   **重點：** 需要熟練掌握「文字題 -> 數學方程 -> 求解 -> 解釋物理意義」的完整流程。特別注意相平面分析在生物模型中的應用。
4. analyse and generate solutions to sometimes complex problems, with logical and coherent methodolgy in solving DEs.
    *   **重點：** 遇到複雜方程時，第一步判斷類型（線性/非線性、常/偏、階數），第二步選擇合適的方法（分離變量、級數、積分變換、定性分析），第三步執行計算，第四步驗證與解釋。

1. First-Order ODEs & Qualitative  

*   **定量求解方法（Quantitative Methods）：**
    *   可分離變量方程、齊次方程。
    *   一階線性方程與積分因子（Integrating factors）。
    *   全微分方程（Exact equations）與伯努利方程（Bernoulli equations）。
*   **存在性與唯一性理論（Theoretical Knowledge）：**
    *   皮卡-林德洛夫定理（Picard-Lindelöf Theorem）與皮卡迭代法。
    *   解對初值和參數的連續依賴性。
*   **定性分析與相平面分析（Qualitative & Phase Plane Analysis）：**
    *   方向場（Direction fields）與等斜線（Isoclines）。
    *   自治系統（Autonomous systems）與一維相線分析。
    *   **二維相平面分析：** 臨界點/平衡點（Equilibrium points）的分類（節點、鞍點、螺旋點、中心點）。
    *   非線性系統的線性化（Linearization）與李雅普諾夫穩定性（Lyapunov stability）基礎。
    *   極限環（Limit cycles）與龐加萊-本迪克松定理（Poincaré-Bendixson Theorem）簡介。

2. Higher-Order Linear ODEs & Series Solutions  

*   **二階及高階線性 ODE 理論：**
    *   齊次與非齊次方程的解的結構。
    *   朗斯基行列式（Wronskian）與阿貝爾定理（Abel's Theorem）。
    *   常係數線性方程的求解（特徵方程法）。
    *   待定係數法（Method of Undetermined Coefficients）與參數變易法（Variation of Parameters）。
*   **冪級數解法（Series Solutions）：**
    *   常點（Ordinary points）處的冪級數解。
    *   正則奇點（Regular singular points）處的 **Frobenius 方法**。
    *   指標方程（Indicial equation）與根的分類（實根、重根、複根）對應的解的形式。
*   **特殊函數（Special Functions - 3年級核心考點）：**
    *   貝塞爾方程（Bessel's equation）與貝塞爾函數。
    *   勒讓德方程（Legendre's equation）與勒讓德多項式。

3. Boundary Value Problems & Sturm-Liouville Theory 連接 PDE 的橋樑

*   **邊值問題（BVPs）基礎：** 初值問題（IVP）與邊值問題（BVP）的區別（解的存在性與唯一性差異）。
*   **施圖姆-劉維爾理論（Sturm-Liouville Theory）：**
    *   正則施圖姆-劉維爾問題（Regular S-L problems）的標準形式。
    *   特徵值（Eigenvalues）與特徵函數（Eigenfunctions）的性質。
    *   特徵函數的正交性（Orthogonality）與權函數（Weight functions）。
*   **傅里葉級數（Fourier Series）：**
    *   傅里葉正弦級數與餘弦級數。
    *   週期函數的傅里葉展開與狄利克雷條件（Dirichlet conditions）。

4. Deep Theory of PDEs * 

    *   二階線性 PDE 的分類：橢圓型（Elliptic）、拋物型（Parabolic）、雙曲型（Hyperbolic）。
    *   特徵線方法（Method of Characteristics）與標準型化簡。
*   **分離變量法（Separation of Variables）：** 將 PDE 轉化為 ODE 邊值問題的系統方法。
*   **三大經典 PDE 的解析解：**
    *   **熱傳導方程（Heat/Diffusion Equation）：** 一維/二維熱傳導，穩態解。
    *   **波動方程（Wave Equation）：** 達朗貝爾解（D'Alembert's solution），行波與駐波，弦振動問題。
    *   **拉普拉斯/泊松方程（Laplace's/Poisson's Equation）：** 穩態溫度分佈，極值原理（Maximum principle）。
*   **積分變換法（Integral Transforms）：**
    *   拉普拉斯變換（Laplace Transform）在求解 PDE 和 ODE 中的應用。
    *   傅里葉變換（Fourier Transform）在無限域 PDE 問題中的應用。

5. Applications & Complex Problem Solving  

    *   機械振動（彈簧-質量-阻尼系統，共振現象）。
    *   電路分析（RLC 電路）。
    *   連續介質力學基礎（弦的橫振動、桿的縱振動推導）。
    *   種群動力學（Population dynamics）。
    *   Lotka-Volterra 捕食者-獵物模型（利用相平面分析其週期性與穩定性）。
    *   如何將複雜的實際物理/工程問題轉化為數學模型（建立 DE）。
    *   當精確解析解無法求得時，如何結合定性分析（如穩定性）和近似/漸近方法來理解系統行為。

linear transformations between vector spaces. This unit offers you an introduction to linear algebra using familiar examples, and explores earlier results, like the Fundamental Theorem of Calculus, in a new light. Building on your understanding of algebra, DEs and multivariable calculus, you will explore systems of linear equations, matrices and algebraic operations on them, and inner products. Learning outcomes:

2. theoretical understanding and computational methods;
    *   **策略：** 考試通常分為兩部分。一部分是純計算（如求 SVD、做 Gram-Schmidt），另一部分是證明（如證明某個映射是線性變換、證明子空間的性質）。**計算要保證 100% 準確，證明要注重邏輯嚴密和定義的準確使用**。
3. prepare for later studies in algebra, statistics, functional analysis, complex variables, differential geometry, signal processing, harmonic analysis, physics; and
    *   **策略：** 這意味著課程會引入一些「無限維」或「抽象」的視角。在學習內積空間時，**一定要重視函數空間（如傅里葉級數的內積定義）的例子**，這是未來學習泛函分析和信號處理的絕對核心。理解「正交投影」在幾何和數據擬合中的雙重意義。
4. illustrate, demonstrate and teach modern mathematics.
    *   **策略：** 這要求你具備「費曼技巧」。在複習時，嘗試用直觀的幾何圖形（如二維/三維空間中的拉伸、旋轉、投影）來解釋抽象的代數概念（如特徵值、核、像）。如果課程有 Presentation 或 Peer-teaching 環節，重點放在**「幾何直觀」與「代數計算」的結合**上。**特別關注「函數空間」和「微積分算子」的線性代數解釋，這是這門課拿高分和真正領悟數學之美的關鍵。**

1. 矩陣代數與線性方程組（具體與計算基礎）*對應大綱：familiar examples, systems of linear equations, matrices and algebraic operations*

*   **線性方程組與矩陣運算：**
    *   高斯消元法（Gaussian elimination）與高斯-若爾當消元法。
    *   矩陣的代數運算（加法、乘法、轉置、逆矩陣）。
    *   行列式（Determinants）的性質與計算（克萊姆法則）。
*   **矩陣的分解與標準型（計算方法）：**
    *   LU 分解（簡介，為後續數值計算做準備）。
    *   矩陣的秩（Rank）、零度（Nullity）與秩-零化度定理（Rank-Nullity Theorem）。

2. 抽象向量空間與線性變換（核心理論）*對應大綱：linear transformations between vector spaces, introduction using familiar examples*

*   **向量空間的公理化定義：**
    *   從熟悉的 $\mathbb{R}^n$ 過渡到抽象向量空間。
    *   **非矩陣向量空間的例子：** 多項式空間 $P_n$、連續函數空間 $C[a,b]$、矩陣空間 $M_{m \times n}$。
*   **子空間與基：**
    *   子空間（Subspaces）、張成（Span）、線性相關與無關。
    *   基（Basis）與維度（Dimension）。
*   **線性變換（Linear Transformations）：**
    *   線性變換的定義、核（Kernel）與像（Image）。
    *   **線性變換的矩陣表示：** 如何在給定基底下將抽象線性變換轉化為矩陣。
    *   基變換（Change of basis）與過渡矩陣（Transition matrices）。
    *   同構（Isomorphism）：證明有限維向量空間與其坐標空間 $\mathbb{R}^n$ 同構。

3. 特徵值理論與微分方程的聯繫（統一性與應用）*對應大綱：unify material learnt in earlier units (DEs), computational methods*

*   **特徵值與特徵向量：**
    *   特徵多項式、特徵值、特徵空間。
    *   矩陣的對角化（Diagonalization）條件與方法。
    *   凱萊-哈密頓定理（Cayley-Hamilton Theorem）。
*   **線性代數在常微分方程（DEs）中的統一應用：**
    *   將高階線性 ODE 轉化為**一階線性 ODE 方程組（狀態空間形式）**。
    *   利用矩陣指數（Matrix exponential, $e^{At}$）和特徵值/特徵向量求解線性常係數 ODE 方程組。

4. 內積空間與正交性（通往高級課程的橋樑）*對應大綱：inner products, prepare for statistics, signal processing, harmonic analysis*

*   **內積空間基礎：**
    *   實/複內積的定義與性質。
    *   範數（Norm）、距離、柯西-施瓦茨不等式（Cauchy-Schwarz Inequality）。
*   **正交性與投影：**
    *   正交補（Orthogonal complements）。
    *   **格拉姆-施密特正交化過程（Gram-Schmidt process）**。
    *   正交投影（Orthogonal projections）與**最小二乘法（Least Squares approximations）**（統計學與數據擬合的核心）。
*   **正交矩陣與酉矩陣：**
    *   正交矩陣（Orthogonal matrices）的性質。
    *   實對稱矩陣的對角化（譜定理 Spectral Theorem 基礎）。

5. 微積分與線性代數的統一（核心亮點）*對應大綱：explores earlier results, like the Fundamental Theorem of Calculus, in a new light; unify multivariable calculus*

*   **微積分算子作為線性變換：**
    *   將微分算子 $D = \frac{d}{dx}$ 和積分算子 $I = \int$ 視為函數空間（如多項式空間 $P_n$ 或光滑函數空間）上的**線性變換**。
    *   求微分算子的矩陣表示（例如在基 $\{1, x, x^2\}$ 下）。
*   **用線性代數重新詮釋微積分基本定理（FTC）：**
    *   從線性代數角度看，FTC 本質上是關於**核（Kernel）與像（Image）** 的關係。
    *   微分算子 $D$ 的核是常數空間。積分算子 $I$ 是 $D$ 的右逆（$D \circ I = \text{Id}$），而 $I \circ D = \text{Id} - P_0$（其中 $P_0$ 是投影到常數空間的投影算子）。這完美解釋了為什麼不定積分要加常數 $C$。
*   **多變量微積分中的線性代數：**
    *   **雅可比矩陣（Jacobian matrix）** 作為多變量函數在某一點的「最佳線性近似（Best linear approximation）」。
    *   海森矩陣（Hessian matrix）與多元函數的極值判定（正定/負定矩陣）。

6. 高級主題與數學表達 *對應大綱：prepare for functional analysis, differential geometry, physics; illustrate and teach modern mathematics*

*   **對偶空間與張量初步（為微分幾何/物理準備）：**
    *   線性泛函（Linear functionals）與對偶空間（Dual space, $V^*$）。
    *   伴隨變換（Adjoint transformations）。
*   **奇異值分解（SVD）與主成分分析（PCA）初步（為統計/信號處理準備）：**
    *   SVD 的幾何意義與計算。
    *   協方差矩陣（Covariance matrices）與數據降維的線性代數基礎。
*   **數學溝通與證明（LO4 的落實）：**
    *   嚴格的數學證明寫作（直接證明、反證法、數學歸納法在線性代數中的應用）。
    *   可能包含小組展示（Presentation）或撰寫數學解釋報告，練習如何「illustrate and teach」數學概念。

    *   整環（Integral domains）與域（Fields）的定義與關係。
    *   子環（Subrings）與理想（Ideals：主理想、素理想、極大理想）。
*   **環的同態與商環：**
    *   環同態、核與像。
    *   **商環（Quotient rings）** 的構造。
    *   **環的同構定理**（與群的同構定理對比學習）。
*   **環的直積：**環的直積結構與中國剩餘定理（Chinese Remainder Theorem）在環論中的推廣。

4. Polynomial Rings & Fields of Quotients *對應大綱：polynomial rings, fields of quotients*

*   **多項式環理論：**
    *   多項式環 $R[x]$ 的性質。
    *   帶餘除法、最大公因式。
    *   不可約多項式（Irreducible polynomials）與愛森斯坦判別法（Eisenstein's criterion）。
*   **特殊整環的分類：**
    *   歐幾里得整環（ED） $\subset$ 主理想整環（PID） $\subset$ 唯一分解整環（UFD） $\subset$ 整環（Integral Domain）。
    *   高斯引理（Gauss's Lemma）與多項式環的 UFD 性質。
*   **商域（Fields of Quotients）：**
    *   從整環構造其商域（例如從 $\mathbb{Z}$ 構造 $\mathbb{Q}$，從多項式環構造有理函數域）。
    *   商域的泛性質（Universal property）。

5. Unification & Forward Look *對應大綱：continue the unification of earlier material commenced in PMTH213; prepare for number theory, algebraic geometry, Lie algebras, etc.*

*   **與 PMTH213（線性代數）的統一：**
    *   **模（Modules）的初步概念：** 將向量空間視為「域上的模」，將阿貝爾群視為「整數環 $\mathbb{Z}$ 上的模」。這將線性代數中的基、維度、線性變換統一到了抽象代數的框架下。
    *   矩陣環 $M_n(F)$ 的結構與線性變換的環同態。
*   **為高級課程鋪路（LO5 的落實）：**
    *   **數論：** 環論中的理想與數論中的整除性。
    *   **代數幾何：** 多項式環的極大理想與仿射空間中的點（希爾伯特零點定理預習）。
    *   **表示論/李代數：** 群作用（Group actions）與對稱性的代數化。
    *   **範疇論/同調代數：** 同態、核、像、同構定理本質上都是範疇論中「態射（morphisms）」的具體體現。

6. Proof Construction & Mathematical Communication  

*   **抽象代數的證明技巧：**
    *   構造性證明（如構造同構映射）。
    *   利用泛性質（Universal properties）進行證明。
    *   反證法與數學歸納法在群/環階數與整除性中的應用。
*   **數學直觀與教學溝通（LO6 的核心）：**
    *   **幾何直觀化：** 例如用正方形的對稱性（翻轉、旋轉）來「illustrate」二面體群 $D_4$；用時鐘算術來「demonstrate」商群 $\mathbb{Z}/n\mathbb{Z}$。
    *   **數學寫作與演講：** 課程可能包含書寫嚴格的證明作業（Proof assignments），甚至要求進行小組展示（Presentation），向同學「teach」某個抽象定理（如西羅定理的證明思路），訓練將極度抽象的概念轉化為他人能聽懂的語言。
