```python
print("\n------ 1.2 量子对易子 ------")
print("使用矩阵表示量子算符（有限维近似）")

def create_matrix_operators(n):
    n = int(n)                       # 强制转为 Python int
    R = CC                           # 复数基域，避免类型歧义
    """创建谐振子升降算符及相关算符的矩阵表示"""
    a = matrix(R, n, n, 0)      # 湮灭算符
    adag = matrix(R, n, n, 0)   # 产生算符
    
    for i in range(n-1):
        a[i, i+1] = sqrt(i+1)
        adag[i+1, i] = sqrt(i+1)
    
    # 位置和动量算符
    x = (a + adag)/sqrt(2)
    p = (a - adag)/(sqrt(2)*I)
    
    return a, adag, x, p

# 创建4维算符
n_dim = 4
a, adag, x_op, p_op = create_matrix_operators(n_dim)

print(f"\n位置算符x ({n_dim}×{n_dim}矩阵):")
print(x_op.n(digits=3))

print(f"\n动量算符p ({n_dim}×{n_dim}矩阵):")
print(p_op.n(digits=3))

# 计算对易子 [x, p]
commutator = x_op*p_op - p_op*x_op
print(f"\n对易子[x, p]:")
print(commutator.n(digits=3))

print(f"\n[x, p]/(iħ)近似单位矩阵:")
quantum_relation = commutator/(I*hbar)
print(quantum_relation)
```

```python
print("\n------ 3.2 量子路径积分 ------")

def path_integral_demo(x_i, x_f, t_i, t_f, N=10):
    """路径积分离散化演示"""
    dt = (t_f - t_i)/N
    times = [t_i + i*dt for i in range(N+1)]
    
    # 经典作用量
    S_classical = m/2 * ((x_f - x_i)^2)/(t_f - t_i)
    
    print(f"\n从 (x={x_i}, t={t_i}) 到 (x={x_f}, t={t_f})")
    print(f"经典作用量 S_cl = {S_classical}")
    
    # 几条示例路径
    print("\n示例路径的作用量计算:")
    paths = []
    for eps in [-0.3, -0.15, 0, 0.15, 0.3]:
        def path_func(t):
            # 线性路径 + 正弦扰动
            linear = x_i + (x_f - x_i)*(t - t_i)/(t_f - t_i)
            perturbation = eps * sin(pi*(t - t_i)/(t_f - t_i))
            return linear + perturbation
        
        # 计算路径的作用量
        S_path = 0
        for i in range(N):
            q_mid = (path_func(times[i]) + path_func(times[i+1]))/2
            q_dot = (path_func(times[i+1]) - path_func(times[i]))/dt
            S_path += lagrangian_harmonic(q_mid, q_dot, times[i]) * dt
        
        phase = exp(I*S_path/hbar)
        paths.append((eps, S_path, phase))
        print(f"ε={eps:5.2f}: S={S_path}, exp(iS/ħ)={phase}")
    
    return paths

print("量子路径积分: K(x_f, t_f; x_i, t_i) = ∫D[x(t)] exp(iS[x(t)]/ħ)")
print("对所有连接初末状态的路径求和")

paths = path_integral_demo(0, 1, 0, 1)
```

# Mathematical Structure Unity: Classical vs Quantum Mechanics

This notebook demonstrates, computationally, the shared mathematical backbone:

- **Classical mechanics:** observables are functions on phase space with a **Poisson bracket**.
- **Quantum mechanics:** observables are operators with a **commutator**.
- **Unity:** both are Lie-algebra dynamics  
  $\dot f = \{f,H\},\qquad \dot{\hat A} = \frac{1}{i\hbar}[\hat A,\hat H].$

## 1) Classical phase space: Poisson bracket + Hamilton equations

```python
# SageMath
var('q p m omega')
R.<q,p> = PolynomialRing(SR, 2)

def poisson(f, g):
    # {f,g} = df/dq dg/dp - df/dp dg/dq
    return diff(f,q)*diff(g,p) - diff(f,p)*diff(g,q)

H = p^2/(2*m) + (m*omega^2*q^2)/2

# Hamilton equations: qdot = {q,H}, pdot = {p,H}
qdot = poisson(q, H)
pdot = poisson(p, H)

s(qdot)
s(pdot)
```

### Jacobi identity check (Poisson bracket is a Lie bracket)

```python
f = q^2*p
g = q*p^2
h = q + p

J = poisson(f, poisson(g,h)) + poisson(g, poisson(h,f)) + poisson(h, poisson(f,g))
s(J)
```

## 2) Quantum: operators + commutators (finite-dimensional truncation)

We model the harmonic oscillator in a truncated Fock space of dimension `N`.
Truncation introduces boundary artifacts near the top state; increasing `N` reduces them.

```python
def annihilation(N):
    # a_{n-1,n} = sqrt(n)
    A = matrix(SR, N, N, 0)
    for n in range(1, N):
        A[n-1, n] = sqrt(n)
    return A

def creation(N):
    return annihilation(N).conjugate_transpose()

def comm(A,B):
    return A*B - B*A

# Parameters
N = 10
var('hbar m omega')

a  = annihilation(N)
ad = creation(N)

# x and p in terms of a, a^\dagger
x = sqrt(hbar/(2*m*omega))*(a + ad)
p_op = sqrt(m*omega*hbar/2)*(1/I)*(a - ad)

# Hamiltonian H = p^2/(2m) + (m omega^2 x^2)/2
H = (p_op*p_op)/(2*m) + (m*omega^2*x*x)/2

# Heisenberg equations: xdot = (1/iħ)[x,H], pdot = (1/iħ)[p,H]
xdot = (1/(I*hbar))*comm(x, H)
pdot = (1/(I*hbar))*comm(p_op, H)

# Compare with expected: xdot ≈ p/m, pdot ≈ -m ω^2 x (up to truncation effects)
test1 = (xdot - p_op/m)
test2 = (pdot + m*omega^2*x)

subs_dict = {hbar:1, m:1, omega:1}

def frob_norm(M):
    return sqrt(sum((M[i,j].subs(subs_dict))^2 for i in range(N) for j in range(N)))

s(frob_norm(test1))
s(frob_norm(test2))
```

## 3) Canonical commutation: dimensionless quadratures

Define
$ Q=\sqrt{\frac{m\omega}{\hbar}}x,\qquad P=\frac{1}{\sqrt{m\omega\hbar}}p.$
In the infinite-dimensional limit, $[Q,P]=i\mathbf{1}$. In truncation it is close to identity except near the boundary.

```python
Q = sqrt(m*omega/hbar)*x
P = (1/sqrt(m*omega*hbar))*p_op

C = comm(Q, P).subs(subs_dict)
C
```

## 4) Deformation quantization view: Moyal bracket (series)

The Moyal bracket is a deformation of the Poisson bracket:
$\{f,g\}_M = \{f,g\} + O(\hbar^2).$
Below we compute the first correction term via a series expansion.

```python
def moyal_bracket_series(f,g):  # {f,g}_M ≈ {f,g} - (hbar^2/24) D^3(f,g) + O(hbar^4)
    D1 = poisson(f,g)
    D3 = (diff(f,q,3)*diff(g,p,3)
          - 3*diff(f,q,2,p,1)*diff(g,p,2,q,1)
          + 3*diff(f,q,1,p,2)*diff(g,p,1,q,2)
          - diff(f,p,3)*diff(g,q,3))

    return (D1 - (hbar^2/24)*D3)

f = q^3
g = p^3

s(poisson(f,g))
s(moyal_bracket_series(f,g))
```
