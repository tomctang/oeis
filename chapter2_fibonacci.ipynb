
# Chapter 2: Fibonacci Numbers

The Fibonacci sequence (A000045) serves as the archetype of linear recursions and exponential growth.  In this chapter we explore the mechanisms behind the Fibonacci numbers as a microcosm of the book’s philosophy.  The sequence appears in multiple mathematical domains—differential equations (DE), linear algebra (MAT), combinatorics and graph theory (GRA), modular arithmetic (MOD), and representation theory (REP)—each shedding light on different aspects of the same numbers.



## Sequence Analyst – Mechanisms & DAG

**Active mechanisms**

- **DE (Primary)** – The Fibonacci numbers satisfy a second‐order linear recurrence $F_n=F_{n-1}+F_{n-2}$, which corresponds to an ordinary differential equation in the generating function’s denominator.  The characteristic polynomial has roots $arphi$ and $\psi$.
- **MAT (Primary)** – The sequence arises from powers of the companion matrix $egin{pmatrix}1&1\1&0\end{pmatrix}$; eigenvalues again reveal the golden ratio.
- **GRA (Secondary)** – Combinatorial interpretations include counting words with no consecutive $b$’s and tilings of a $1	imes n$ board.
- **MOD (Secondary)** – Reduction of the sequence modulo $m$ exhibits periodicity; these periods (Pisano periods) reveal arithmetic structure.
- **REP (Secondary)** – The sequence counts certain free monoid words and can be interpreted via representations of the free monoid on two generators.

**Base–Bridge–Emergence DAG**

- **Base:** Start with the combinatorial definition: $F_n$ counts the number of ways to tile a $1	imes n$ board with $1	imes1$ and $1	imes2$ tiles or the number of binary strings of length $n$ without consecutive $1$’s.  This discrete counting problem anchors the sequence.
- **Bridge:** Formulate the linear recurrence $F_{n}=F_{n-1}+F_{n-2}$ and express it through matrix multiplication and eigenvalues.  The bridge connects discrete counts to linear algebra and differential equations.
- **Emergence:** From the recurrence derive generating functions and closed forms.  The generating function $F(x)=x/(1-x-x^2)$ encodes the entire sequence analytically, and partial fraction decomposition uncovers the role of the golden ratio.  Modular reductions exhibit periodicity, hinting at deeper number‐theoretic structure.

The DAG flows from combinatorial foundations to algebraic representations and finally to analytic expressions.  Each arrow represents a mechanism transforming one viewpoint into another.  In code, this will guide the order of demonstrations.


```

print("="*70)
print("FIBONACCI NUMBERS")
print("="*70)

# --- 1. Direct Computation ---
print("
1. First 20 Fibonacci numbers:")
# Compute the first 20 Fibonacci numbers directly using the built-in function
fibs = [fibonacci(n) for n in range(20)]
print(fibs)
```

```

# --- 2. ODE / RECURRENCE RELATION ---
print("
2. ODE CONNECTION: y'' = y' + y")
print("   Characteristic equation: λ^2 = λ + 1")
print("   Solutions: λ = (1 ± sqrt(5))/2 (golden ratio φ and conjugate ψ)")

phi = (1 + sqrt(5))/2
psi = (1 - sqrt(5))/2

print(f"
   Golden ratio φ = {phi.n(digits=10)}")
print(f"   Conjugate ψ = {psi.n(digits=10)}")

def fibonacci_binet(n):
    '''Binet formula: F_n = (φ^n - ψ^n)/√5'''
    phi = (1 + sqrt(5))/2
    psi = (1 - sqrt(5))/2
    return ((phi^n - psi^n)/sqrt(5)).simplify_full()

print("
   Via Binet's formula (closed form from ODE):")
fibs_binet = [fibonacci_binet(n) for n in range(10)]
print(f"   {fibs_binet}")

# Verify it matches the built-in function
print(f"   Match? {all(fibonacci_binet(n) == fibonacci(n) for n in range(15))}")
```

```

# --- 3. MATRIX POWERS ---
print("
3. MATRIX: Powers of [[1,1],[1,0]]")

F = matrix([[1, 1], [1, 0]])
print("   Fibonacci matrix F = ")
print(F)

print("
   Powers of F:")
for n in range(1, 8):
    Fn = F^n
    print(f"   F^{n} = {Fn.list()}, F_{n} = {Fn[0,1]}, F_{n+1} = {Fn[0,0]}")

# Eigenvalues are φ and ψ
print("
   Eigenvalues of F:")
eigenvals = F.eigenvalues()
print(f"   {[e.n(digits=10) for e in eigenvals]}")
print(f"   These are φ and ψ!")

# Diagonalization: F = PDP⁻¹
print("
   Diagonalization F = PDP⁻¹:")
P, D = F.eigenmatrix_right()
print(f"   P = {P}")
print(f"   D = {D}")
print(f"   Verify: F = PDP⁻¹? {F == P*D*P^(-1)}")
```

```

# --- 4. GENERATING FUNCTION ---
print("
4. GENERATING FUNCTION: F(x) = x/(1 - x - x^2)")

R.<x> = PowerSeriesRing(QQ, default_prec=20)
F_gen = x/(1 - x - x^2)
print(f"   F(x) = {F_gen}")
print(f"   Coefficients: {F_gen.list()}")

# Verify partial fractions (from ODE eigenvalues)
print("
   Partial fraction decomposition (using φ, ψ):")
print("   F(x) = 1/√5 * [1/(1-φx) - 1/(1-ψx)]")

# Reset variable x for symbolic series expansion
x = var('x')
phi = (1 + sqrt(5))/2
psi = (1 - sqrt(5))/2

F_partial = (1/sqrt(5)) * ( 1/(1 - phi*x) - 1/(1 - psi*x) )
print(F_partial.series(x, 15))  # expand as a power series up to x^14
```

```

# --- 5. CONTINUED FRACTIONS ---
print("
5. CONTINUED FRACTIONS: φ = [1; 1, 1, 1, 1, ...]")

cf = continued_fraction(phi)
print(f"   Golden ratio as continued fraction: {cf}")
print(f"   First 15 convergents:")

for i in range(15):
    conv = cf.convergent(i)
    print(f"   p_{i}/q_{i} = {conv} = {conv.numerator()}/{conv.denominator()}", end="")
    if i > 0:
        print(f" → Ratio of F_{i+1}/F_{i}")
    else:
        print()
```

```

# --- 6. GROUP THEORY / WORD STRUCTURES ---
print("
6. GROUP THEORY: Fibonacci counts in free monoid")
print("   Words of length n using {a, b} with no consecutive b's")

def count_no_consecutive_b(n, memo={0: 1, 1: 2}):
    '''Count binary strings of length n with no consecutive b's'''
    if n in memo:
        return memo[n]
    # If ends with 'a' (0) or 'b' (1)
    result = count_no_consecutive_b(n-1, memo) + count_no_consecutive_b(n-2, memo)
    memo[n] = result
    return result

counts = [count_no_consecutive_b(n) for n in range(12)]
print(f"   Counts: {counts}")
print(f"   These are F_{{n+2}}!")
```

```

# --- 7. TILINGS ---
print("
7. TILINGS: Ways to tile 1×n board with 1×1 and 1×2 tiles")

def count_tilings(n, memo={0: 1, 1: 1}):
    '''Count ways to tile a 1×n board'''
    if n in memo:
        return memo[n]
    result = count_tilings(n-1, memo) + count_tilings(n-2, memo)
    memo[n] = result
    return result

tilings = [count_tilings(n) for n in range(15)]
print(f"   Tiling counts: {tilings}")
print("   These are Fibonacci numbers!")
```

```

# --- 8. MODULAR ARITHMETIC (Pisano Periods) ---
print("
8. MODULAR ARITHMETIC: Fibonacci mod m (Pisano periods)")

def pisano_period(m):
    '''Find period of Fibonacci sequence mod m'''
    prev, curr = 0, 1
    for i in range(m * m):
        prev, curr = curr, (prev + curr) % m
        if prev == 0 and curr == 1:
            return i + 1
    return -1

print("   Pisano periods π(m) for small m:")
for m in range(2, 13):
    period = pisano_period(m)
    fib_mod = [fibonacci(n) % m for n in range(period)]
    print(f"   π({m:2d}) = {period:3d}, sequence mod {m}: {fib_mod[:min(12, period)]}")
```


## Mathematical Expositor – Conceptual Explanations

1. **Direct computation:** Listing the first few terms grounds the reader in the concrete sequence.  Patterns such as the growth rate and initial conditions become evident.

2. **ODE connection and Binet’s formula:** Solving the linear recurrence leads to a characteristic polynomial whose roots are the golden ratio $arphi$ and its conjugate $\psi$.  Binet’s formula expresses $F_n$ as a closed form $(arphi^n - \psi^n)/\sqrt{5}$, revealing the exponential growth rate and the pervasive appearance of $arphi$.

3. **Matrix powers:** Viewing the recurrence as repeated multiplication by the companion matrix connects the sequence to linear algebra.  The eigenvalues and diagonalization of the matrix mirror the ODE solutions and demonstrate how matrix powers encode the sequence.

4. **Generating function:** The rational generating function $F(x)=x/(1-x-x^2)$ packages all terms of the sequence into a single analytic object.  Partial fraction decomposition recovers the same eigenvalues $arphi$ and $\psi$, showing unity between algebraic and analytic perspectives.

5. **Continued fractions:** The continued fraction expansion of the golden ratio is $[1;1,1,1,\dots]$, and its convergents are ratios of consecutive Fibonacci numbers.  This ties Diophantine approximation to the recurrence and highlights why $arphi$ is optimally approximated by Fibonacci ratios.

6. **Group theory / word structures:** Counting binary strings with no consecutive $1$’s (or $b$’s) yields the Fibonacci numbers.  This interpretation comes from the free monoid on two generators and shows how combinatorial restrictions produce the same recurrence.

7. **Tilings:** Tiling a $1	imes n$ board with $1	imes1$ and $1	imes2$ tiles is another manifestation of the recurrence: a tiling either ends with a monomino or a domino, leading to the same recursive sum.

8. **Modular arithmetic:** Reducing Fibonacci numbers modulo $m$ yields periodic sequences with period $\pi(m)$, the Pisano period.  Studying these periods uncovers rich arithmetic structure and hints at connections to number theory and cryptography.

These explanations emphasize how disparate mechanisms converge to produce the same sequence, illustrating the core theme of the book: sequences as interfaces between mathematical worlds.
