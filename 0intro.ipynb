## Calculus - Finite Difference - Combinatorics 

排列组合是“离散的微积分”，排列组合提供了组合恒等式和离散的单项，而有限差分微积分提供了一套处理离散序列的计算工具

帕斯卡三角形不仅是组合数的列表，它也是第 n 阶差分为0的函数表，体现了差分方程的性质。 

牛顿-莱布尼茨公式在离散情况下对应“组合数求和”公式 

How permutations enter directly?
- The permutation number $P(x,n) = \frac{x!}{(x−n)!} = x^{\underline{n}}$  (when x ≥ n integer) So:
- $P(x,n)  ≡  x$ permute $n$
- $Δ P(x,n) = n ⋅ P(x, n−1)$
- This is the discrete analogue of $\frac{d}{dx} x^n = n x^{n−1}$

How combinations enter directly?
- The binomial coeff $\binom{x}{n} = \frac{x(x−1)⋯(x−n+1)}{n!} = \frac{P(x,n)}{n!}$, So $Δ \binom{x}{n} = \binom{x}{n−1} $
- This is one of the most beautiful (and useful) identities in discrete calculus (x can be any real or complex number for generalized binomial coeff)
